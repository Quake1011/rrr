```
import pandas as pd     #подключение модуля pandas
import sqlite3      #подключение модуля для работы с локальными БД sqlite
con = sqlite3.connect("data.sqlite") #  подключаемся к локальной базе 'data.sqlite'
# con = sqlite3.connect(C:/Users/tseve/Desktop/data.sqlite") с рабочего стола
# доступны три таблицы: 
#     lvl_base
#     lvl_base_geoip
#     lvl_base_hits
df = pd.read_sql_query("select * from lvl_base ORDER BY value DESC", con)     #выполняем запрос на вывод данных, отсортированных по колонке value
print(df)     #выводим рез-т
```

```
df = pd.read_sql_query("SELECT *, \
(SELECT region FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS region,\
(SELECT clientip FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS clientip,\
(SELECT country FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS country,\
(SELECT city FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS city,\
(SELECT country_code FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS country_code \
FROM lvl_base WHERE steam = 'STEAM_1:1:180196110'",con)     #выполняем запрос по уникальному steamid
print(df)     #выводим рез-т
```

```
df = pd.read_sql_query("SELECT *, \
(SELECT region FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS region,\
(SELECT clientip FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS clientip,\
(SELECT country FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS country,\
(SELECT city FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS city,\
(SELECT country_code FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS country_code \
FROM lvl_base ORDER BY value DESC LIMIT 10",con)     #выполняем запрос на получения данных первых 10-ти человек, отсортированных по убыванию
print(df)     #выводим рез-т
```

```
df = pd.read_sql_query("SELECT *, \
(SELECT region FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS region,\
(SELECT clientip FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS clientip,\
(SELECT country FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS country,\
(SELECT city FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS city,\
(SELECT country_code FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS country_code \
FROM lvl_base where value > 3000000",con)     #выполняем запрос на получения данных значение которых больше 3000000
print(df)     #выводим рез-т
```

```
df = pd.read_sql_query("SELECT steam, \
(SELECT country FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS country,\
(SELECT country_code FROM lvl_base_geoip WHERE steam = lvl_base.steam) AS country_code \
FROM lvl_base where steam='STEAM_1:1:180196110'",con)     #выполняем запрос на получение страны и ее кода по steamid
print(df)     #выводим рез-т
```
